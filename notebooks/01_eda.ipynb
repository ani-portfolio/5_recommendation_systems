# Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Import Data

```python
df_movies = pd.read_csv("/Users/ani/Projects/5_recommendation_system/data/df_movies_processed.csv")
df_movies.head(5)
```

# Plots

```python
df_movies.describe()
```

```python
# Rating Distribution
cols = ['vote_average', 'vote_count', 'popularity', 'runtime', 'budget', 'revenue']

plt.subplots(3, 2, figsize=(15, 10))

for i, col in enumerate(cols):
    plt.subplot(3, 2, i + 1)
    plt.hist(df_movies[col].dropna(), bins=50, alpha=0.7)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
cols = ['revenue', 'popularity']

fig, axes = plt.subplots(1, 2, figsize=(20, 5))
axes = axes.flatten()  # Convert 2D array to 1D for easy indexing

for i, col in enumerate(cols):
    axes[i].scatter(df_movies['budget'], df_movies[col], alpha=0.5)
    axes[i].set_title(f'Budget vs {col}')
    axes[i].set_xlabel('Budget')
    axes[i].set_ylabel(col)
    axes[i].set_xscale('log')
    axes[i].set_yscale('log')

plt.tight_layout()
plt.show()
```
